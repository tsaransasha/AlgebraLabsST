###  3. Символічні обчислення та алгебра виразів
а) оголосіть змінні $x=10$ і $y=15$ та обчисліть їхню суму і добуток;<br>
б) оголосіть символічну змінну $z$ і обчисліть вираз $3z^2 + 2z - 5$;<br>
в) спростіть вираз $2x^2 + 4x - 2x^2 + x$;<br>
г) розкрийте дужки у виразі $(x + 2)(x - 3)$;<br>
д) розкладіть на множники тричлен $x^2 - 9x + 20$;<br>
е) розв'яжіть рівняння $x^2 - 4x + 3 = 0$.<br>

```sage
# Ваш код тут:# а) x = 10, y = 15, їхня сума і добуток
x = 10
y = 15

print("Сума:", x + y)
print("Добуток:", x * y)

# б) вираз 3z^2 + 2z - 5
z = var('z')
expr = 3*z^2 + 2*z - 5
print(expr)

# в) спрощення виразу
expr = 2*x^2 + 4*x - 2*x^2 + x
print("Спрощений вираз:", simplify(expr))

# г) розкриття дужок
expr = (x + 2)*(x - 3)
print("Після розкриття дужок:", expand(expr))

# д) розклад на множники
expr = x^2 - 9*x + 20
print("Розклад на множники:", factor(expr))

# е) розв'язання рівняння
x = var('x')
equation = x^2 - 4*x + 3 == 0
print(solve(equation, x))
```

### 4. Похідні, інтеграли та графіки функцій

а) побудуйте графік функції $f(x) = x^2 - 4x + 3$ на відрізку $[-5, 5]$; <br>
б) знайдіть похідну функції $f(x) = x^3 + 5x^2 - 2x + 7$; <br>
в) знайдіть невизначений інтеграл функції $2x^3 - x^2 + 5$; <br>
г) обчисліть визначений інтеграл $\int_{1}^{4} x^2\, dx$; <br>
д) оголосіть функцію $f(x) = 3x^2 - 2x + 1$ та обчисліть $f(5)$; <br>
е) побудуйте на одній площині графіки $f(x)=x^2$ і $g(x)=-x^2+5$ на $[-3,3]$.


```sage
# а) графік f(x) = x^2 - 4x + 3 на [-5, 5]
x = var('x')
f = x^2 - 4*x + 3

show(plot(f, (x, -5, 5)))

# б) похідна
f = x^3 + 5*x^2 - 2*x + 7
print("Похідна:", diff(f, x))

# в) невизначений інтеграл
f = 2*x^3 - x^2 + 5
print("Невизначений інтеграл:", integral(f, x))

# г) визначений інтеграл від 1 до 4
f = x^2
print("Визначений інтеграл:", integral(f, x, 1, 4))

# д) функція f(x) = 3x^2 - 2x + 1 та f(5)
f(x) = 3*x^2 - 2*x + 1
print("f(5) =", f(5))

# е) два графіки на одній площині
f(x) = x^2
g(x) = -x^2 + 5

p1 = plot(f(x), (x, -3, 3))
p2 = plot(g(x), (x, -3, 3))

show(p1 + p2)
```

### 5. Програмування
а) напишіть умовний оператор, який визначає, чи число $n$ додатне, від'ємне або дорівнює нулю; <br>
б) напишіть цикл, що виводить числа від 1 до 10; <br>
в) напишіть функцію, яка перевіряє чи є задане число $n$ простим.

```sage
# Ваш код тут:# а) визначення знаку числа

n = 7

if n > 0:
    print("Число додатне")
elif n < 0:
    print("Число від'ємне")
else:
    print("Число дорівнює нулю")


# б) числа від 1 до 10

for i in range(1, 11):
    print(i)


# в) перевірка, чи є число простим

def is_prime(n):
    if n < 2:
        return False
    
    for i in range(2, n):
        if n % i == 0:
            return False
    
    return True


n = 17

if is_prime(n):
    print(n, "— просте число")
else:
    print(n, "— не є простим числом")
```
