### 2. Працюємо з елементами груп
Для кожної з груп: $\mathbb{Z}_{10}$, $\mathbb{Z}_{10}^{*}$, $A_4$, $GL_2(\mathbb{Z}_3)$, виконайте наступні кроки:
1. Побудуйте групу.
2. Виберіть довільні елементи $a, b \in G$ та обчисліть добуток $a \cdot b$.
3. Знайдіть обернені елементи $a^{-1}, b^{-1}$ та виконайте перевірку правильності.
4. Побудуйте підгрупу, породжену елементами $a$ та $b$, і визначте її порядок.
5. Знайдіть всі елементи групи, які є оберненими самі до себе, тобто $a^{-1}=a$.
6. Якщо група $G$ не є абелевою, знайдіть такі елементи $a, b \in G$, що $a \cdot b \neq b \cdot a$.

```sage
# Ваш код тут:
print(" Дослідження групи Z_10")

# 1. 
G = CyclicPermutationGroup(10)
print("1. Група побудована:", G)

# 2. 
elements = list(G)
a = elements[1]
b = elements[3]
product = a * b
print(f"2. Вибрані елементи: a = {a}, b = {b}")
print(f"   Добуток a * b = {product}")

# 3. 
inv_a = a^-1
inv_b = b^-1
print(f"3. Обернений елемент a^-1 = {inv_a} (перевірка a * a^-1 = {a * inv_a})")
print(f"   Обернений елемент b^-1 = {inv_b} (перевірка b * b^-1 = {b * inv_b})")

# 4. 
subgroup = G.subgroup([a, b])
print(f"4. Породжена підгрупа H = : {subgroup}")
print(f"   Порядок підгрупи: {subgroup.order()}")

# 5. 
self_inverse_elements = [x for x in G if x^-1 == x]
print(f"5. Елементи, які обернені самі до себе: {self_inverse_elements}")

# 6. 
print("6. Дослідження на комутативність:")
if not G.is_abelian():
    found = False
    for x in G:
        for y in G:
            if x * y != y * x:
                print(f"   Знайдено комутуючі/некомутуючі: x = {x}, y = {y}, x*y = {x*y}, y*x = {y*x}")
                found = True
                break
        if found:
            break
else:
    print("   Група Z_10 є абелевою (комутативною), тому всі елементи переставні (a * b = b * a для всіх).")
```

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
# Ваш код тут:
print(" Дослідження підгруп групи Z_24 ")

# 1. 
G = CyclicPermutationGroup(24)
print("1. Група побудована:", G)

# 2. 
subgroups = G.subgroups()
print(f"2. Загальна кількість підгруп: {len(subgroups)}")
print("   Список підгруп:", subgroups)

# 3. 
isomorphism_classes = [H.structure_description() for H in subgroups]
unique_classes = set(isomorphism_classes)
print(f"3. Кількість підгруп з точністю до ізоморфізму: {len(unique_classes)}")
print("   Класи ізоморфізму:", unique_classes)

# 4. 
orders = [H.order() for H in subgroups]
print(f"4. Порядки всіх підгруп: {orders}")

# 5.
group_order = G.order()
lagrange_check = all(group_order % h_ord == 0 for h_ord in orders)
print(f"5. Теорема Лагранжа виконується (порядок кожної підгрупи ділить порядок групи {group_order}): {lagrange_check}")
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

Для кожної з груп: $D_4$, $Q_8$, $S_3$, $S_4$, $A_4$, виконайте наступні кроки:
1. Побудуйте граф Келі для деякої системи твірних групи.
2. Визначте степінь вершин графа Келі.
3. Перевірте, чи граф Келі є вершинно-транзитивним.
4. Дослідіть, чи граф Келі містить гамільтонів цикл.

```sage
print("=== ГРАФ КЕЛІ ДЛЯ D4 ===")
D4 = DihedralGroup(4)
print(f"Порядок групи: {D4.order()}")
S = D4.gens()
print(f"Система твірних: {S}")
cg = D4.cayley_graph(generators=S)
print(f"Граф Келі має {cg.num_verts()} вершин та {cg.num_edges()} ребер")
print(f"Степінь вершин графа: {set(cg.degree(v) for v in cg.vertices())}")
print(f"Чи є вершинно-транзитивним?: {cg.is_vertex_transitive()}")
print(f"Чи містить гамільтонів цикл?: {cg.is_hamiltonian()}")
plot = cg.plot(color_by_label=True, vertex_size=500, vertex_labels=True, vertex_colors={'skyblue': cg.vertices()})
plot.show()
print("\n" + "="*50 + "\n")
```
